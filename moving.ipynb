```python
from pathlib import Path
import shutil
```

```python
source_dir = Path(r"z:\Buzsakilabspace\LabShare\NoamNitzan\Open_Access\Allen_2022\data")
sessions = os.listdir(r"F:\behavior_ecephys_sessions")
for session in sessions:
    session_path = source_dir / f'session_{session}'
    if int(session) in [1069461581, 1099598937, 1099869737, 1104289498, 1109889304]: continue 
    if session_path.exists():
        print(f"{session}")
        lfp_path_src = source_dir / f'session_{session}' / 'probeC_lfp.nwb'
        lfp_path_dst = Path(f"F:/behavior_ecephys_sessions/{session}/probe_probeC_lfp.nwb")
        if not lfp_path_src.exists():
            print(f"LFP file for session {session} does not exist at the source path.")
            continue
        if not lfp_path_dst.exists():
            shutil.copy(lfp_path_src, lfp_path_dst)
        else:
            print(f"LFP file already exists at the dest path.")
    else:
        print(f"Session {session} not found in source directory.")
```
